```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def plot_snapshot(filename, title=None):
    data = np.loadtxt(filename)

    x = data[:,0]
    y = data[:,1]
    theta = data[:,2]

    u = np.cos(theta)
    v = np.sin(theta)

    plt.figure(figsize=(4,4))
    plt.quiver(
        x, y, u, v,
        angles='xy',
        scale_units='xy',
        scale=1,
        width=0.008,
        color='black'
    )

    plt.axis('equal')
    plt.axis('off')

    if title:
        plt.title(title)

    plt.tight_layout()
    plt.show()
```

```python
etas = [0.2, 0.4, 0.6, 0.8]

for eta in etas:
    filename = f"snapshot_eta_{int(eta*10)}.dat"
    plot_snapshot(filename, title=f"η = {eta}")
```
