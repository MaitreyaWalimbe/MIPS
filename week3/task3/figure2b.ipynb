```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt("phi_vs_rho.dat")

rho = data[:, 0]
phi = data[:, 1]
```

```python
plt.figure(figsize=(5,4))
plt.plot(rho, phi, 'o-', lw=2)

plt.xlabel(r"Density $\rho$")
plt.ylabel(r"Order parameter $\Phi$")
plt.title(r"Vicsek model: $\Phi$ vs $\rho$ ($\eta=0.6$)")

plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
