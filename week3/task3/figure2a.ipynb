```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
N_values = [40, 100, 400, 4000]

plt.figure(figsize=(6,5))

for N in N_values:
    filename = f"phi_vs_eta_N{N}.dat"
    data = np.loadtxt(filename)

    eta = data[:,0]
    phi = data[:,1]

    plt.plot(eta, phi, marker='o', linewidth=1.5, label=f"N={N}")
plt.xlabel(r"Noise strength $\eta$")
plt.ylabel(r"Order parameter $\Phi$")
plt.legend(frameon=False)
plt.tight_layout()
plt.show()
```
